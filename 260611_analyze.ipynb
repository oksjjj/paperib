```python
import os
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
```

```python
# data 디렉토리 내 파일 목록
files = os.listdir('data')

# CSV 파일만 선택
csv_files = [f for f in files if f.endswith('.csv')]

# dataframe들을 담을 리스트
df_list = []

# 파일별로 읽기
for file in csv_files:
    path = os.path.join('data', file)
    temp_df = pd.read_csv(path)
    df_list.append(temp_df)

# 하나의 dataframe으로 병합
df = pd.concat(df_list, ignore_index=True)

df = df.sort_values(['time', 'PLMN'])

df['time'] = pd.to_datetime(df['time'])

df
```

```python
df.info()
```

```python
# PLMN별 M971 개수 계산
result = (
    df.groupby("PLMN")["M971"]
    .sum()
    .reset_index(name="M971_sum")
    .sort_values("M971_sum", ascending=False)
)

# CSV 파일로 저장
result.to_csv("analysis/plmn_m971_sum_sorted.csv", index=False)
```

```python
# 저장 폴더 생성
SAVE_DIR = "graph/trend"
os.makedirs(SAVE_DIR, exist_ok=True)

# 파일 번호
save_idx = 1


def visualize(PLMN):

    global save_idx

    df_temp = df[
        df['PLMN'] == PLMN
    ].copy()

    # 제외할 컬럼
    exclude_cols = ['time', 'PLMN']

    # plot 대상 컬럼
    plot_cols = [
        c for c in df_temp.columns
        if c not in exclude_cols
    ]

    # plot
    ax = df_temp.plot(
        x='time',
        y=plot_cols,
        figsize=(20, 8),
        legend=False
    )

    plt.xticks(rotation=45)
    plt.tight_layout()

    # 예: 001_45005.png
    filename = f"{save_idx:03d}_{PLMN}.png"

    plt.savefig(
        os.path.join(SAVE_DIR, filename),
        dpi=150,            # 적당한 해상도
        bbox_inches='tight'
    )

    plt.close()

    save_idx += 1
```

```python
for plmn in result['PLMN'][:800]:
    print(f"========== {plmn} ==========")
    visualize(plmn)
```
